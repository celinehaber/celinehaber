```python
# exercice 1
# Création de la liste de nombres entiers de 1 à 10
liste = list(range(1, 11))

# Ajout de 11 à la fin de la liste
liste.append(11)

# Insertion de 0 au début de la liste
liste.insert(0, 0)

# Suppression de l'élément au milieu de la liste
index_milieu = len(liste) // 2
liste.pop(index_milieu)

# Affichage de la liste finale
print("Liste finale :", liste)
```

```python
# exercice 2
mots = ["Python", "est", "amusant", "et", "puissant"]

print("Tous les mots :")
for mot in mots:
    print(mot)

print("\nMots contenants plus de 4 lettres :")
for mot in mots:
    if len(mot) > 4:
        print(mot)
```

```python
#exercice 3
def addition(a, b):
    return a + b


def multiplication(liste):
    produit = 1
    for nombre in liste:
        produit *= nombre
    return produit

somme = addition(7, 3)
print(f"La somme de 7 et 3 est {somme}")

liste_nombres = [2, 3, 4]
produit_total = multiplication(liste_nombres)
print(f"Le produit des nombres {liste_nombres} est {produit_total}")
```

```python
#exercice 4
def moyenne(liste):
    if len(liste) == 0:
        return "La liste est vide, impossible de calculer sa moyenne."
    return sum(liste) / len(liste)

# Test de la fonction avec une liste d'exemples
exemple_liste = [18, 25, 30, 41, 62]
resultat_moyenne = moyenne(exemple_liste)

# Affichage du résultat
print(f"La moyenne de {exemple_liste} est {resultat_moyenne}")
```

```python
#exercice 5
import numpy as np

tableau = np.arange(10)

# Calcul des statistiques
somme = np.sum(tableau)
moyenne = np.mean(tableau)
maximum = np.max(tableau)
minimum = np.min(tableau)

# Affichage des résultats
print(f"Tableau : {tableau}")
print(f"Somme : {somme}")
print(f"Moyenne : {moyenne}")
print(f"Maximum : {maximum}")
print(f"Minimum : {minimum}")

# Multiplication de chaque élément par 2
tableau_double = tableau * 2
print(f"Tableau après multiplication par 2 : {tableau_double}")
```

```python
#exercice 6
import numpy as np

tableau = np.random.rand(10)

somme = np.sum(tableau)
moyenne = np.mean(tableau)

# Tableau en ordre croissant
tableau_trie = np.sort(tableau)

# Affichage des résultats
print(f"Tableau : {tableau}")
print(f"Somme : {somme}")
print(f"Moyenne : {moyenne}")
print(f"Tableau trié : {tableau_trie}")
```

```python
# exercice 7
import matplotlib.pyplot as plt

x = list(range(1, 11))  # Liste de nombres de 1 à 10
y = [i**2 for i in x]   # Liste des carrés de chaque nombre

# Création du graphique
plt.plot(x, y, marker='o', linestyle='-', color='c')

# Ajout d'un titre et des étiquettes
plt.title("Graphique des carrés des nombres")
plt.xlabel("Valeurs de x")
plt.ylabel("Valeurs de y = x²")


# Affichage du graphique
plt.grid(False)
plt.show()
```

```python
# exercice 8
import numpy as np
import matplotlib.pyplot as plt

# Génération de 1000 valeurs aléatoires suivant une distribution normale (0;1)
valeurs = np.random.normal(loc=0, scale=1, size=1000)

plt.hist(valeurs, bins=25, color='blue', edgecolor='black', alpha=0.5)

plt.title("Histogramme des valeurs générées (Distribution Normale)")
plt.xlabel("Valeurs")
plt.ylabel("Fréquence")

# Affichage
plt.grid(True)
plt.show()
```

```python
# exercice 9
import numpy as np

# Création d'un tableau 5x5 initialisé à zéro
tableau = np.zeros((5, 5), dtype=int)

for i in range(tableau.shape[0]):  # Parcours des lignes
    for j in range(tableau.shape[1]):  # Parcours des colonnes
        tableau[i, j] = i + j

print("Tableau généré avec la somme des indices :")
print(tableau)
```

```python
# exercice 10
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)

# Calcul des valeurs de sin(x) et cos(x)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin, label="sin(x)", color="blue", linestyle="-")
plt.plot(x, y_cos, label="cos(x)", color="red", linestyle="--")

plt.title("Courbes de sin(x) et cos(x)")
plt.xlabel("x (radians)")
plt.ylabel("Valeurs de sin(x) et cos(x)")


plt.legend()
plt.show()
```
